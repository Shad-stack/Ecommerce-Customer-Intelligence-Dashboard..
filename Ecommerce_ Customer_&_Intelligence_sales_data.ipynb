```python
# 1. Changing Object to Datetime

df['order_date'] = pd.to_datetime(df['order_date'], format='mixed', dayfirst =True)


df['ship_date'] = pd.to_datetime(df['ship_date'], format='mixed', dayfirst=True)
```

```python
# 2. checking null value

df.isnull().sum()
```

```python
# 3. find Duplicate value

df.duplicated().sum()
```

```python
df.head(2)
```

```python
# 4. Profit Margin

df['profit_margin-pct'] = (df['profit'] / df['sales'] *100)
```

```python
# 5.Monthly sales 

Monthly_sales = df.groupby('order_month')['sales'].sum().reset_index()

# Growth percentage (%)
Monthly_sales['growth_pct'] = Monthly_sales['sales'].pct_change() * 100
Monthly_sales.round(2)
```

```python
# Monthly Sales aggregate karna
monthly_sales = df.groupby('order_month')['sales'].sum()

# Line plot draw karna
plt.figure(figsize=(10, 4))
monthly_sales.plot(kind='line', marker='o', color='b')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

```python
df.head(2)
```

```python
# 6. category analysis

df.groupby('category')['sales'].sum().sort_values(ascending=False)
```

```python
# 6.1. Category-wise Total Sales, Profit & Quantity

category_summary = df.groupby('category').agg({
    'order_id': 'count',
    'sales': 'sum',
    'profit': 'sum',
    'quantity': 'sum'
}).rename(columns={'order_id': 'total_orders'}).reset_index()
category_summary
```

```python
# 6.2.  Category-wise Average Profit Margin %

category_margin = df.groupby('category').agg({
    'sales': 'sum',
    'profit': 'sum'
}).reset_index()

# Exact Category Profit Margin Formula
category_margin['profit_margin_%'] = (category_margin['profit'] / category_margin['sales']) * 100
category_margin.round(2)
```

```python
# 6.3. Category % in total sales

category_sales = df.groupby('category')['sales'].sum().reset_index()
category_sales['sales_share_%'] = (category_sales['sales'] / df['sales'].sum()) * 100
category_sales.sort_values(by='sales_share_%', ascending=False).round(2)
```

```python
# 6.4. Graph of category sales

plt.figure(figsize=(8, 4))
df.groupby('category')['sales'].sum().sort_values(ascending=False).plot(kind='bar', color='skyblue')
plt.title('Total Sales by Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
# 7. Correlation in all dataset

df.corr(numeric_only=True).round(2)
```

```python
# 7.1.  correlation between profit & sales

df['sales'].corr(df['profit']).round(2)
```

```python
# 7.2.  Correlation graph

plt.figure(figsize=(8, 6))
# Heatmap draw karna
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

```python
# 8. Distribution

# Sales Column Ka Distribution Chart
plt.figure(figsize=(8, 4))
sns.histplot(df['sales'], kde=True, color='blue')
plt.title('Sales Distribution')
plt.xlabel('Sales Amount')
plt.ylabel('Frequency (Count)')
plt.show()
```
